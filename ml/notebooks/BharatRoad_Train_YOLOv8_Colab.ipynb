# BharatRoad Twin — YOLOv8 Road Damage Detection Training (Google Colab)

This notebook trains a custom YOLOv8 model for detecting potholes, cracks, and damaged road patches on Indian roads.

## 1. Setup & Environment Verification

```python
!pip install -q ultralytics roboflow pyyaml
import torch
from ultralytics import YOLO
print(f"PyTorch Version: {torch.__version__}")
print(f"CUDA Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"Device Name: {torch.cuda.get_device_name(0)}")
```

## 2. Train YOLOv8 Small Model on Dataset

```python
model = YOLO('yolov8s.pt')
results = model.train(
    data='dataset/data.yaml',
    epochs=50,
    imgsz=640,
    batch=16,
    name='bharatroad_yolov8s',
    project='runs/detect'
)
```

## 3. Evaluate Precision, Recall, and mAP@50

```python
metrics = model.val()
print(f"mAP@50: {metrics.box.map50:.4f}")
print(f"mAP@50-95: {metrics.box.map:.4f}")
```

## 4. Export Model Weights to ONNX / TorchScript

```python
model.export(format='onnx')
```
